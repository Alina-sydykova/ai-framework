## Очистка

```python
df_clean = df.drop_duplicates().copy()
print("shape after drop_duplicates", df_clean.shape)
```

```python
tip_median = df_clean['tip'].median()
print("tip median: ", tip_median)

df_clean['tip'] = df_clean['tip'].fillna(tip_median)
```

```python
rating_median = df_clean['rating'].median()
print("rating median: ", rating_median)

df_clean['rating'] = df_clean['rating'].fillna(rating_median)
```

```python
df_clean = df_clean.drop(['rating'], axis = 1)
```

```python
print(df_clean.isna().sum())
```

## 5. Анализы: фильтры и groupby

Несколько коротких вопросов к данным — по одной идее на ячейку.

### 5.1. Средний чек у членов программы лояльности и остальных


```python
df_clean.groupby('is_member')['price'].mean().round(2)
```

### 5.2. Какие напитки берут вечером и ночью?


```python
df_clean[df_clean['time_slot'].isin(['evening', 'late_night'])]['drink'].value_counts()
```

### 5.3. Среднее время приготовления по типу еды


```python
df_clean.groupby('food_item')['prep_minutes'].mean().round(2).sort_values(ascending=False)
```

### 5.4. Дождь и горячий шоколад — совпадение?


```python
(df_clean['drink'] == 'hot_chocolate').groupby(df_clean['raining']).mean().round(3)
```

### 5.5. Курс студента и средний рейтинг / чаевые


```python
df_clean.groupby('student_year')[['rating', 'tip']].mean().round(2)
```

### 5.6. Факультет и калории (кто «слаще» заказывает?)


```python
df_clean.groupby('faculty')['calories'].mean().round(1).sort_values(ascending=False)
```

## 6. Графики

Подписываем оси и заголовки. Без лишнего декора. Каждую диаграмму следет построить через plt и seaborn


### 6.1. Гистограмма цены

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df_clean, x='price', bins=20)
plt.title('Распределение цены заказа')
plt.xlabel('Цена')
plt.ylabel('Количество заказов')
plt.show()
```

### 6.2. Столбчатая диаграмма: заказы по дням недели


```python
order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
plt.figure(figsize=(8, 5))
sns.countplot(data=df_clean, x='day_of_week', order=[d for d in order if d in df_clean['day_of_week'].unique()])
plt.title('Заказы по дням недели')
plt.xlabel('День недели')
plt.ylabel('Количество заказов')
plt.show()
```

### 6.3. Средняя цена по напитку (barh удобнее читать)


```python
avg_price = df_clean.groupby('drink')['price'].mean().sort_values()
plt.figure(figsize=(8, 5))
sns.barplot(x=avg_price.values, y=avg_price.index, orient='h')
plt.title('Средняя цена по напитку')
plt.xlabel('Средняя цена')
plt.ylabel('Напиток')
plt.show()
```

### 6.4. Scatter: время приготовления vs рейтинг


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x='prep_minutes', y='rating', alpha=0.6)
plt.title('Время приготовления и рейтинг')
plt.xlabel('Время приготовления (мин)')
plt.ylabel('Рейтинг')
plt.show()
```

### 6.5. Boxplot калорий по размеру стакана


```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_clean, x='size', y='calories', order=['S', 'M', 'L'])
plt.title('Калории по размеру стакана')
plt.xlabel('Размер')
plt.ylabel('Калории')
plt.show()
```

### 6.6. Связь учёбы и кофе: study hours vs prior cups (scatter + лёгкий цвет по member)

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x='study_hours_today', y='prior_coffee_cups', hue='is_member', alpha=0.7)
plt.title('Часы учёбы и выпитый кофе')
plt.xlabel('Часов учёбы сегодня')
plt.ylabel('Чашек кофе до заказа')
plt.show()
```

### Задание A

Посчитайте **средний `tip`** для каждого способа оплаты (`payment`).  
Отсортируйте по убыванию среднего чаевого. Постройте bar chart.


```python
tip_by_payment = df_clean.groupby('payment')['tip'].mean().sort_values(ascending=False)
print(tip_by_payment.round(3))

plt.figure(figsize=(8, 5))
sns.barplot(x=tip_by_payment.index, y=tip_by_payment.values)
plt.title('Средние чаевые по способу оплаты')
plt.xlabel('Способ оплаты')
plt.ylabel('Средние чаевые')
plt.show()
```
